## Object Detection using Pre-trained SSD MobileNetV2

This notebook follows `DL_Assi_6.pdf`: load a pre-trained SSD MobileNetV2 model from TensorFlow Hub, read an image, preprocess it, detect objects, draw bounding boxes for detections above 0.5 confidence, and display the result.

## 2. Import libraries

```python
import tensorflow as tf
import tensorflow_hub as hub
import cv2
import matplotlib.pyplot as plt
```

## 3. Load the pre-trained SSD MobileNetV2 model

```python
model = hub.load("https://tfhub.dev/tensorflow/ssd_mobilenet_v2/2")
print("SSD MobileNetV2 model loaded successfully")
```

## 4. Load the input image

Place `testm.jpeg` in the notebook's working directory. In Google Colab, you can uncomment and run the upload lines to select an image from your computer.

```python
# For Google Colab, uncomment these lines to upload an image:
# from google.colab import files
# uploaded = files.upload()

image_path = 'testm.jpeg'  # Change this if your image has a different name
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(
        f'Could not read {image_path}. Upload the image or correct image_path.'
    )

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis('off')
plt.title('Input Image')
plt.show()
```

## 5. Preprocess the image

```python
input_image = tf.convert_to_tensor(image, dtype=tf.uint8)
input_image = tf.expand_dims(input_image, axis=0)

print("Input tensor shape:", input_image.shape)
```

## 6. Perform object detection

```python
result = model(input_image)
print("Detection completed")
print("Available result keys:", list(result.keys()))
```

## 7. Get detection results

```python
boxes = result["detection_boxes"][0].numpy()
classes = result["detection_classes"][0].numpy()
scores = result["detection_scores"][0].numpy()

print("Number of detections returned:", len(scores))
```

## 8. Define labels and draw bounding boxes

```python
h, w, _ = image.shape

labels = {
    1: "person",
    2: "bicycle",
    3: "car",
    4: "motorcycle",
    5: "airplane",
    6: "bus",
    7: "train",
    8: "truck",
    9: "boat",
    17: "cat",
    18: "dog",
    19: "horse",
    20: "sheep",
    21: "cow"
}

for i in range(len(scores)):
    if scores[i] > 0.5:
        y1, x1, y2, x2 = boxes[i]

        x1, x2 = int(x1 * w), int(x2 * w)
        y1, y2 = int(y1 * h), int(y2 * h)

        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

        class_id = int(classes[i])
        label = labels.get(class_id, "Object")
        cv2.putText(
            image, label, (x1, max(y1 - 10, 0)),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2
        )

print("Bounding boxes drawn for detections with confidence > 0.5")
```

## 9. Display the detected image

```python
plt.figure(figsize=(10, 8))
plt.imshow(image)
plt.axis('off')
plt.title('Object Detection using SSD MobileNetV2')
plt.show()
```

## Result

The pre-trained SSD MobileNetV2 model detects objects in the input image. Objects with a confidence score greater than 0.5 are displayed with green bounding boxes and class labels.
